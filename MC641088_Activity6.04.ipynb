```python
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline

data = pd.read_csv("../smartphone_sales.csv")

data.head()
```

```python
plt.figure(figsize=(12, 6))

plt.stackplot(
    data["Quarter"],
    data["Apple"],
    data["Samsung"],
    data["Huawei"],
    data["Xiaomi"],
    data["OPPO"],
    labels=["Apple", "Samsung", "Huawei", "Xiaomi", "OPPO"]
)

plt.xlabel("Quarter")
plt.ylabel("Sales")
plt.title("Smartphone Sales by Brand Over Time")
plt.legend(loc="upper left")

plt.tight_layout()
plt.show()
```
